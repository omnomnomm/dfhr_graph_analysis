```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt(
    "../analysis/rmsd_rmsf/S99N/rmsd_ref0.xvg",
    comments=["#", "@"]
)

print(data.shape)
print(data[:5])
```

```python
time = data[:, 0]
rmsd = data[:, 1]

print(f"Time range: {time.min()} to {time.max()}")
print(f"RMSD range: {rmsd.min():.3f} to {rmsd.max():.3f} nm")
```

```python
plt.figure(figsize=(10, 4))
plt.plot(time, rmsd, linewidth=0.5)
plt.xlabel("Time (ns)")
plt.ylabel("RMSD (nm)")
plt.title("S99N backbone RMSD")
plt.grid(alpha=0.3)
plt.show()
```

```python
data = np.loadtxt(
    "../analysis/rmsd_rmsf/WT/rmsd_ref0.xvg",
    comments=["#", "@"]
)
time, rmsd = data[:, 0], data[:, 1]

plt.figure(figsize=(10, 4))
plt.plot(time, rmsd, linewidth=0.5)
plt.xlabel("Time (ns)"); plt.ylabel("RMSD (nm)")
plt.title("WT backbone RMSD")
plt.grid(alpha=0.3); plt.show()
```

```python
d = np.loadtxt("../analysis/rmsd_rmsf/I155L/rmsd_ref0.xvg", comments=["#", "@"])

plt.figure(figsize=(14, 4))
plt.plot(d[:, 0], d[:, 1], linewidth=0.4)
plt.xlabel("Time (ns)")
plt.ylabel("RMSD (nm)")
plt.title("I155L backbone RMSD")
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(14, 4))
for name in ["WT", "S99N", "I155L"]:
    d = np.loadtxt(f"../analysis/rmsd_rmsf/{name}/rmsd_ref0.xvg", comments=["#", "@"])
    plt.plot(d[:, 0], d[:, 1], linewidth=0.4, alpha=0.7, label=name)
plt.xlabel("Time (ns)"); plt.ylabel("RMSD (nm)")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.show()
```
