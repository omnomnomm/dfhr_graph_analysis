# Protein Contact Network Construction — PfDHFRBuilds the occupancy-based contact network for one or more variants.The chain, following the formal definitions document:| Symbol | Shape | Object ||---|---|---|| `X` | (T, N, 3) | Cα coordinates || `D(t)` | (N, N) | distances, per frame || `S` | (N, N) | sequence-separation mask || `C(t)` | (N, N) | contact indicator, per frame || `W` | (N, N) | **occupancy matrix** || `A`, `Ã` | (N, N) | adjacency / weighted adjacency || `G` | — | **protein contact network** |Set `VARIANTS` in the config cell to choose what to run.

---## 2. `X` — the coordinate tensorReads the trajectory and keeps only the 219 Cα atoms, discarding water and all otheratoms. Result has shape **(T, N, 3)**: frames x residues x (x, y, z).Cached to `.npy` — the slow read off the external drive happens once per variant.Re-running this cell afterwards loads from SSD in under a second.

```python
def extract_ca(variant, force=False):    """Return (xyz_nm, resnums). Cached on disk."""    spec    = VARIANT_SPEC[variant]    xyz_out = f"{OUTDIR}/{variant}_ca_xyz.npy"    res_out = f"{OUTDIR}/{variant}_resnums.npy"    if os.path.exists(xyz_out) and not force:        xyz, resnums = np.load(xyz_out), np.load(res_out)        print(f"{variant}: cached  {xyz.shape}")        return xyz, resnums    top_path  = f"{ROOT}/{spec['top']}"    traj_path = f"{ROOT}/{spec['traj']}"    for p in (top_path, traj_path):        if not os.path.exists(p):            raise FileNotFoundError(f"{variant}: missing {p}  (external drive mounted?)")    top = md.load_topology(top_path)    ca  = top.select("protein and name CA")    if len(ca) != N_RES:        raise ValueError(f"{variant}: found {len(ca)} CA atoms, expected {N_RES}")    resnums = np.array([top.atom(i).residue.resSeq for i in ca])    t0, chunks = time.time(), []    for chunk in md.iterload(traj_path, chunk=2000, top=top, atom_indices=ca):        chunks.append(chunk.xyz)        print(f"  {variant}: {sum(c.shape[0] for c in chunks)} frames", end="\r", flush=True)    xyz = np.concatenate(chunks).astype(np.float32)    np.save(xyz_out, xyz); np.save(res_out, resnums)    print(f"\n{variant}: {xyz.shape}  resSeq {resnums[0]}-{resnums[-1]}  "          f"{time.time()-t0:.0f}s")    return xyz, resnumscoords = {v: extract_ca(v) for v in VARIANTS}
```

---## 4. `S` — the sequence-separation maskResidues adjacent in the chain are close regardless of how the protein folds, so theyrecord backbone geometry rather than fold architecture.A distance floor alone does not remove them. Consecutive Cα sit at ~3.8 A and areexcluded by the 4 A cutoff, but in an α-helix the i,i+2 and i,i+3 separations are~5.4 A and ~5.0 A — both inside the contact band. They must be excluded by**sequence position**, not distance.$$S_{ij} = \begin{cases} 1 & |i-j| > s \\ 0 & \text{otherwise}\end{cases}$$

```python
idx = np.arange(N_RES)SEP = np.abs(np.subtract.outer(idx, idx))      # |i-j| for every pairMASK = SEP > MIN_SEP                           # Siu = np.triu_indices(N_RES, k=1)               # each pair oncen_possible = MASK[iu].sum()print(f"all pairs            : {len(iu[0])}")print(f"allowed by mask      : {n_possible}  ({100*n_possible/len(iu[0]):.1f}%)")print(f"excluded by mask     : {len(iu[0]) - n_possible}")fig, ax = plt.subplots(figsize=(4, 3.4))ax.imshow(MASK[:40, :40], cmap="Greys_r", origin="lower")ax.set_title(f"mask S (first 40 residues), s = {MIN_SEP}", fontsize=10)ax.set_xlabel("residue"); ax.set_ylabel("residue")plt.tight_layout(); plt.show()
```

---## 5. `D(t)` and `C(t)` — one frame at a time**`D(t)`** holds the Euclidean distance between every pair of Cα in a single frame.**`C(t)`** is the binary verdict for that frame: 1 if the pair is inside the contactband *and* allowed by the mask, 0 otherwise.Neither is stored. They are computed inside the loop and discarded — only the runningcount survives. Shown here on a single frame to make the objects concrete.

```python
frame = xyz[0] * 10.0                                        # nm -> angstromD = np.linalg.norm(frame[:, None, :] - frame[None, :, :], axis=-1)C = (D >= CUTOFF_LO) & (D <= CUTOFF_HI) & MASKprint(f"D shape {D.shape}, symmetric: {np.allclose(D, D.T)}, diagonal {D[0,0]:.1f}")print(f"C shape {C.shape}, contacts in this frame: {C[iu].sum()}")fig, ax = plt.subplots(1, 2, figsize=(10, 4.2))im0 = ax[0].imshow(D, cmap="magma_r", origin="lower", vmax=40)ax[0].set_title("D(t=0)  —  distances (A)"); plt.colorbar(im0, ax=ax[0], shrink=0.8)ax[1].imshow(C, cmap="Greys", origin="lower")ax[1].set_title(f"C(t=0)  —  contacts ({CUTOFF_LO}-{CUTOFF_HI} A, |i-j|>{MIN_SEP})")for a in ax: a.set_xlabel("residue"); a.set_ylabel("residue")plt.tight_layout(); plt.show()
```

---## 6. `W` — the occupancy matrixAverage the per-frame verdicts over the window. Each entry is the **fraction of thewindow** during which that pair was in contact.$$W_{ij} = \frac{1}{T}\sum_{t=1}^{T} C_{ij}(t)$$Binary in, continuous out. This is what distinguishes the approach from building anetwork off a single structure: a contact present 95% of the time and one present 30%of the time are now different objects.

```python
def occupancy_matrix(xyz_window, cutoff_lo=CUTOFF_LO, cutoff_hi=CUTOFF_HI, mask=MASK):    """W for one window. xyz_window: (T, N, 3) in nm."""    xa = xyz_window * 10.0    counts = np.zeros((xa.shape[1], xa.shape[1]))    for fr in xa:        d = np.linalg.norm(fr[:, None, :] - fr[None, :, :], axis=-1)        counts += (d >= cutoff_lo) & (d <= cutoff_hi) & mask    return counts / xa.shape[0]def split_windows(xyz, dt_ps, n_windows=N_WINDOWS):    """Slice into n_windows of WINDOW_NS each, using the variant's own timestep."""    w = frames_per_window(dt_ps)    if xyz.shape[0] < n_windows * w:        raise ValueError(f"trajectory has {xyz.shape[0]} frames, "                         f"need {n_windows*w} for {n_windows} x {WINDOW_NS} ns")    return [xyz[k*w:(k+1)*w] for k in range(n_windows)]
```

```python
# demonstrate on one windowwins = split_windows(xyz, VARIANT_SPEC[VARIANTS[0]]["dt_ps"])print(f"{len(wins)} windows of {wins[0].shape[0]} frames")t0 = time.time()W_demo = occupancy_matrix(wins[0])print(f"W computed in {time.time()-t0:.0f}s   shape {W_demo.shape}")print(f"range [{W_demo.min():.2f}, {W_demo.max():.2f}]   "      f"distinct values: {len(np.unique(W_demo))}  <- continuous, not binary")fig, ax = plt.subplots(1, 2, figsize=(11, 4.2))im = ax[0].imshow(W_demo, cmap="viridis", origin="lower", extent=[1, N_RES, 1, N_RES])ax[0].set_title("W — occupancy, window 1"); plt.colorbar(im, ax=ax[0], shrink=0.8)ax[0].set_xlabel("residue"); ax[0].set_ylabel("residue")vals = W_demo[iu]ax[1].hist(vals[vals > 0], bins=50, color="#4472C4")ax[1].axvline(THRESHOLD, color="crimson", ls="--", label=f"theta = {THRESHOLD}")ax[1].set_yscale("log"); ax[1].set_xlabel("occupancy"); ax[1].set_ylabel("pairs")ax[1].set_title("distribution of non-zero occupancy"); ax[1].legend()plt.tight_layout(); plt.show()
```

---## 7. Choosing `theta`Sweep the threshold and watch how the network responds. A **smooth** curve means thetopology is not an artefact of where the line was drawn — that is a robustnessargument worth being able to make.A sharp cliff would mean the opposite, and the threshold would become a parameter todefend rather than a convenience.

```python
def threshold_scan(W, resnums, thresholds=np.arange(0.90, 0.20, -0.05)):    rows = []    for th in thresholds:        A = (W >= th)        G = nx.from_numpy_array(np.triu(A, k=1).astype(int))        G = nx.relabel_nodes(G, {i: int(r) for i, r in enumerate(resnums)})        giant = len(max(nx.connected_components(G), key=len))        rows.append({"theta": float(th), "edges": G.number_of_edges(),                     "giant": giant, "isolated": len(list(nx.isolates(G))),                     "ridge_in": sum(1 for r in RIDGE if G.degree(r) > 0),                     "mut_in": sum(1 for r in MUT_SITES if G.degree(r) > 0)})    return rowsscan = threshold_scan(W_demo, resnums)print(f"{'theta':>6} {'edges':>6} {'giant':>6} {'iso':>4} {'ridge':>6} {'mut':>4}")for r in scan:    print(f"{r['theta']:>6.2f} {r['edges']:>6} {r['giant']:>6} {r['isolated']:>4} "          f"{r['ridge_in']:>4}/11 {r['mut_in']:>2}/2")fig, ax = plt.subplots(1, 2, figsize=(10, 3.6))th = [r["theta"] for r in scan]ax[0].plot(th, [r["edges"] for r in scan], "o-"); ax[0].set_ylabel("edges")ax[1].plot(th, [r["giant"] for r in scan], "o-"); ax[1].set_ylabel("giant component")for a in ax:    a.axvline(THRESHOLD, color="crimson", ls="--"); a.set_xlabel("theta"); a.grid(alpha=0.3)plt.tight_layout(); plt.show()
```

---## 8. `A`, `Ã` and the network `G`Thresholding `W` gives the adjacency matrix. Retaining the occupancy value on survivingpairs gives the weighted adjacency.$$A_{ij} = \mathbb{1}[W_{ij} \ge \theta] \qquad \tilde{A} = W \odot A$$The graph is the same object in network notation. In practice it is built straight from`W` — `A` and `Ã` are never materialised, though either is recoverable from `G`.**Node labels are residue numbers**, not array indices, so node 164 is residue 164.

```python
def graph_from_occupancy(W, resnums, theta=THRESHOLD):    """G = (V, E, w) from the occupancy matrix."""    G = nx.Graph()    G.add_nodes_from(int(r) for r in resnums)    for i, j in zip(*np.triu_indices(len(resnums), k=1)):        if W[i, j] >= theta:            G.add_edge(int(resnums[i]), int(resnums[j]), weight=float(W[i, j]))    return GG_demo = graph_from_occupancy(W_demo, resnums)print(f"G: {G_demo.number_of_nodes()} nodes, {G_demo.number_of_edges()} edges")print(f"   mean degree {2*G_demo.number_of_edges()/N_RES:.2f}")print(f"   density     {nx.density(G_demo):.4f}")# the matrices, recovered from the graphA_bin = nx.to_numpy_array(G_demo, weight=None)A_wgt = nx.to_numpy_array(G_demo, weight="weight")print(f"\nA  entries: {np.unique(A_bin)}")print(f"A~ non-zero range: [{A_wgt[A_wgt>0].min():.2f}, {A_wgt[A_wgt>0].max():.2f}]"      f"   <- bounded below by theta")
```

---## 9. Build all windows, all selected variantsThreshold is held fixed across windows and variants — comparison is only meaningfulif the construction is identical everywhere.

```python
def build_variant(variant, theta=THRESHOLD, n_windows=N_WINDOWS):    xyz, resnums = coords[variant]    dt = VARIANT_SPEC[variant]["dt_ps"]    wins = split_windows(xyz, dt, n_windows)    graphs, occs, iso = {}, {}, {}    print(f"\n{variant}  ({wins[0].shape[0]} frames/window, theta={theta})")    for k, w in enumerate(wins, start=1):        W = occupancy_matrix(w)        G = graph_from_occupancy(W, resnums, theta)        graphs[k], occs[k] = G, W        iso[k] = {int(n) for n in nx.isolates(G)}        giant = len(max(nx.connected_components(G), key=len))        print(f"  w{k:02d}: {G.number_of_edges():4d} edges  giant {giant:3d}/{N_RES}  "              f"<k>={2*G.number_of_edges()/N_RES:5.2f}  iso {len(iso[k]):2d}  "              f"ridge {sum(1 for r in RIDGE if G.degree(r)>0):2d}/11  "              f"mut {sum(1 for r in MUT_SITES if G.degree(r)>0)}/2")    always, ever = set.intersection(*iso.values()), set.union(*iso.values())    print(f"  isolated in ALL windows  ({len(always)}): {sorted(always)}")    print(f"  isolated in SOME window  ({len(ever-always)}): {sorted(ever-always)}")    out = f"{OUTDIR}/{variant}_occupancy_graphs.pkl"    with open(out, "wb") as f:        pickle.dump({"variant": variant, "graphs": graphs, "occupancy": occs,                     "resnums": resnums, "theta": theta,                     "isolated_always": always, "isolated_ever": ever,                     "params": {"cutoff_lo": CUTOFF_LO, "cutoff_hi": CUTOFF_HI,                                "min_sep": MIN_SEP, "window_ns": WINDOW_NS,                                "frames_per_window": wins[0].shape[0],                                "dt_ps": dt}}, f)    print(f"  saved -> {out}")    return graphs, occsresults = {v: build_variant(v) for v in VARIANTS}
```
